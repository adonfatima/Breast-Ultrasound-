```python
import sys, glob
import numpy as np

# fake_shifts.py khuje ber kora
found = glob.glob("/kaggle/input/**/fake_shifts.py", recursive=True)
print("Found:", found)
assert found, "fake_shifts.py paoa jaini. Dataset-e file ache kina dekho."

sys.path.append(found[0].rsplit("/", 1)[0])
from fake_shifts import *

# nokol image ar mask
yy, xx = np.mgrid[:512, :512]
mask = ((yy - 250) ** 2 / 60 ** 2 + (xx - 260) ** 2 / 90 ** 2) < 1
img = np.random.default_rng(0).random((512, 512)) * 0.5 + mask * 0.3

# image shifts
for name in IMAGE_SHIFTS:
    out, _ = apply_shift(img, mask, image_shift=name, strength=0.5)
    print("image", name, out.shape, round(float(out.mean()), 3))

# mask shifts
for name in MASK_SHIFTS:
    scores = []
    for s in (0.0, 0.5, 1.0):
        _, m = apply_shift(img, mask, mask_shift=name, strength=s)
        scores.append(round(float(dice(mask, m)), 3))
    print("mask", name, scores)
```
